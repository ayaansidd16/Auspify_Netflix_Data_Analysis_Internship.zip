```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Task_1_Data_Cleaning/output/Netflix_Cleaned.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Release Year Data Type:", df["release_year"].dtype)

print("Missing Release Years:", df["release_year"].isnull().sum())

print("Oldest Release Year:", df["release_year"].min())

print("Latest Release Year:", df["release_year"].max())

print("Unique Release Years:", df["release_year"].nunique())
```

```python
release_year_counts = (
    df["release_year"]
    .value_counts()
    .sort_index()
)

print("Year-wise Content Releases:")
display(release_year_counts)
```

```python
highest_year = release_year_counts.idxmax()
highest_count = release_year_counts.max()

lowest_year = release_year_counts.idxmin()
lowest_count = release_year_counts.min()

print("Year with Highest Releases:", highest_year)
print("Highest Number of Releases:", highest_count)

print("\nYear with Lowest Releases:", lowest_year)
print("Lowest Number of Releases:", lowest_count)
```

```python
# Netflix Content Releases by Release Year

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate yearly content releases
release_year_counts = (
    df["release_year"]
    .value_counts()
    .sort_index()
)

# Create figure
plt.figure(figsize=(13, 6))

# Netflix Red line chart
plt.plot(
    release_year_counts.index,
    release_year_counts.values,
    color="#E50914",
    marker="o",
    markersize=5,
    linewidth=2.5,
    markerfacecolor="#E50914",
    markeredgecolor="white",
    markeredgewidth=0.7
)

# Title and labels
plt.title(
    "Netflix Content Releases by Release Year",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)

# Styling
plt.grid(
    True,
    linestyle="--",
    color="grey",
    alpha=0.3
)

sns.despine()
plt.tight_layout()
plt.show()
```

```python
# Netflix Content Release Trends (2000–2021)

import matplotlib.pyplot as plt
import seaborn as sns

# Filter release years from 2000 to 2021
recent_releases = df[
    (df["release_year"] >= 2000) &
    (df["release_year"] <= 2021)
]

# Calculate yearly releases
yearly_counts = (
    recent_releases["release_year"]
    .value_counts()
    .sort_index()
)

# Create figure
plt.figure(figsize=(13, 6))

# Netflix Red line chart
plt.plot(
    yearly_counts.index,
    yearly_counts.values,
    color="#E50914",
    marker="o",
    markersize=6,
    linewidth=2.5,
    markerfacecolor="#E50914",
    markeredgecolor="white",
    markeredgewidth=0.8
)

# Title and labels
plt.title(
    "Netflix Content Release Trends (2000–2021)",
    fontsize=16,
    fontweight="bold",
    pad=15
)

plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)

# Styling
plt.xticks(range(2000, 2022, 2))
plt.grid(
    True,
    linestyle="--",
    color="grey",
    alpha=0.3
)

sns.despine()
plt.tight_layout()
plt.show()
```

### Key Business Insights

1. **Gradual Growth:** From 2000 to 2014, Netflix's available content showed a gradual increase in release years.

2. **Rapid Expansion:** A significant increase was observed between 2015 and 2016, indicating a growing presence of recent content in the dataset.

3. **Peak Year:** 2018 recorded the highest number of titles, with 1,146 releases.

4. **Declining Trend:** After 2018, the number of titles decreased, reaching 592 in 2021.

5. **Content Strategy:** The distribution suggests that the dataset contains a strong concentration of titles released during the 2015–2020 period.

6. **Business Observation:** Netflix can use release-year patterns to understand its content library's age distribution and identify periods with higher content representation.

### Conclusion

The analysis shows that Netflix's content library contains titles from 1925 to 2021. Content releases increased significantly after 2010 and reached their peak in 2018. A decline was observed from 2019 to 2021. These findings help understand the release-year distribution and historical patterns of Netflix content.
