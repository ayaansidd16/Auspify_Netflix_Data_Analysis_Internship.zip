```python
import pandas as pd
import numpy as np

print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)

print("Netflix Data Cleaning Project Started!")
print("Python environment is ready.")
```

```python
# Import the Netflix dataset

df = pd.read_csv("dataset/Dataset.csv")

# Display first 5 rows
df.head()
```

```python
# Check dataset dimensions

print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())
```

```python
# Get dataset information

df.info()
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python
# Check unique values in Type column

print("Content Types:")
print(df["type"].value_counts())
```

```python
# Check unique rating categories

print("Rating Categories:")
print(df["rating"].value_counts())
```

```python
# Check unique countries

print("Total Unique Countries:", df["country"].nunique())

print("\nCountry Names:")
print(df["country"].unique())
```

```python
# Check leading and trailing spaces

columns_to_check = ["type", "rating", "country"]

for col in columns_to_check:
    spaces = df[col].astype(str).ne(
        df[col].astype(str).str.strip()
    ).sum()

    print(f"{col}: {spaces} values with extra spaces")
```

```python
# Check unique values after trimming spaces

for col in columns_to_check:
    print(f"\n{col.upper()}")

    print(
        df[col].astype(str).str.strip().value_counts().head(10)
    )
```

```python
# Check placeholder values in the dataset

for col in ["type", "rating", "country"]:

    print(f"\nChecking {col.upper()} column:")

    print(
        df[col][
            df[col].astype(str).str.strip().str.lower().isin(
                ["not given", "unknown", "n/a", "na", "null", ""]
            )
        ].value_counts()
    )# Check placeholder values in the dataset

for col in ["type", "rating", "country"]:

    print(f"\nChecking {col.upper()} column:")

    print(
        df[col][
            df[col].astype(str).str.strip().str.lower().isin(
                ["not given", "unknown", "n/a", "na", "null", ""]
            )
        ].value_counts()
    )
```

```python
# Create a copy of original dataset

df_clean = df.copy()

# Standardize text columns by removing extra spaces

text_columns = df_clean.select_dtypes(include="str").columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

print("Text formatting standardized successfully!")
```

```python
# Replace placeholder values in Country column

df_clean["country"] = df_clean["country"].replace(
    "Not Given", "Unknown"
)

# Clean spaces around individual country names

df_clean["country"] = df_clean["country"].apply(
    lambda x: ", ".join(country.strip() for country in x.split(","))
)

print("Country column standardized successfully!")

print("\nUnknown Country Records:",
      (df_clean["country"] == "Unknown").sum())
```

```python
print("Content Types:")
print(df_clean["type"].value_counts())

print("\nRating Categories:")
print(df_clean["rating"].value_counts())
```

```python
# Final dataset validation

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate Records:")
print(df_clean.duplicated().sum())

print("\nUnknown Country Records:")
print((df_clean["country"] == "Unknown").sum())
```

```python
# Export cleaned dataset

df_clean.to_csv(
    "output/Netflix_Cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully!")
```
